```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

def add_random_purchase_date(
    df,
    start='2024-01-01',
    end='2026-06-30',
    date_col='Ticket_Purchase_Date',
    datetime_col='Ticket_Purchase_Date_dt',
    year_col='Ticket_Purchase_Year',
    month_col='Ticket_Purchase_Month'
):
    """
    Adds random ticket purchase dates and derived year/month columns to df.

    - start, end: strings 'YYYY-MM-DD' defining the date range.
    - date_col: name of the string date column (DD-MM-YYYY).
    - datetime_col: name of the pandas datetime column.
    - year_col, month_col: names for extracted year and month.
    """

    # 1. Convert start/end to datetime
    start_dt = datetime.strptime(start, '%Y-%m-%d')
    end_dt = datetime.strptime(end, '%Y-%m-%d')

    # 2. Compute the range in days
    delta_days = (end_dt - start_dt).days

    # 3. Generate random day offsets
    n_rows = len(df)
    random_days = np.random.randint(0, delta_days + 1, size=n_rows)

    # 4. Build random datetime values
    random_datetimes = [start_dt + timedelta(days=int(d)) for d in random_days]

    # 5. Store as DD-MM-YYYY strings
    df[date_col] = [dt.strftime('%d-%m-%Y') for dt in random_datetimes]

    # 6. Convert to pandas datetime
    df[datetime_col] = pd.to_datetime(df[date_col], format='%d-%m-%Y')

    # 7. Extract year and month
    df[year_col] = df[datetime_col].dt.year
    df[month_col] = df[datetime_col].dt.month

    return df
```

```python
import re
import pandas as pd

def clean_name_column(df, name_col='name', new_col='name_clean'):
    """
    Clean a Titanic-style Name column by:
    - Removing common titles (Mr, Mrs, Miss, Ms, Master, Dr, etc.)
    - Converting to lowercase
    - Stripping extra whitespace

    df:       pandas DataFrame
    name_col: column name with raw names
    new_col:  column name to store cleaned names
    """

    # Work on a string view to be safe
    names = df[name_col].astype(str)

    # 1. Define common titles in a regex pattern
    titles_pattern = r'\b(Mr|Mrs|Miss|Ms|Master|Dr|Rev|Col|Major|Capt|Lady|Sir|Mlle|Mme|Countess)\.?'

    # 2. Remove titles using regex
    #    - re.IGNORECASE handles upper/lower/mixed case
    names_no_titles = names.apply(
        lambda x: re.sub(titles_pattern, '', x, flags=re.IGNORECASE)
    )

    # 3. Remove extra spaces and commas around the remaining name
    names_cleaned = names_no_titles.str.replace(',', ' ', regex=False)
    names_cleaned = names_cleaned.str.replace('\s+', ' ', regex=True)  # collapse multiple spaces
    names_cleaned = names_cleaned.str.strip()

    # 4. Convert to lowercase
    df[new_col] = names_cleaned.str.lower()

    return df
```

```python
import re
import pandas as pd

def clean_name_column(df, name_col='name', new_col='name_clean'):
    """
    Clean a Titanic-style Name column by:
    - Removing common titles (Mr, Mrs, Miss, Ms, Master, Dr, etc.)
    - Converting to lowercase
    - Stripping extra whitespace

    df:       pandas DataFrame
    name_col: column name with raw names
    new_col:  column name to store cleaned names
    """

    # Work on a string view to be safe
    names = df[name_col].astype(str)

    # 1. Define common titles in a regex pattern
    titles_pattern = r'\b(Mr|Mrs|Miss|Ms|Master|Dr|Rev|Col|Major|Capt|Lady|Sir|Mlle|Mme|Countess)\.?'

    # 2. Remove titles using regex
    names_no_titles = names.apply(
        lambda x: re.sub(titles_pattern, '', x, flags=re.IGNORECASE)
    )

    # 3. Remove extra spaces and commas
    names_cleaned = names_no_titles.str.replace(',', ' ', regex=False)
    names_cleaned = names_cleaned.str.replace(r'\s+', ' ', regex=True)  # collapse multiple spaces
    names_cleaned = names_cleaned.str.strip()

    # 4. Convert to lowercase and save as new column
    df[new_col] = names_cleaned.str.lower()

    return df

```
